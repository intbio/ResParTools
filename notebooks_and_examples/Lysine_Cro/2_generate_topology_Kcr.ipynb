```python
def path_dict(path_list):
    mol_dict = {} 
    if isinstance(path_list, str):
        name = os.path.basename(path_list).split('.')[0]
        mol_dict[name] = path_list
    if isinstance(path_list, list):
        for path in path_list:
            name = os.path.basename(path).split('.')[0]
            mol_dict[name] = path
    return mol_dict
```
